 ## Pandas Tutorial 

```python
import pandas as pd 
```

## How to Frame Data

```python
myData = {
    'cars':["Porshe", "Mercedez", "Mclaren"], 
    'books':["IT", "To Kill A Mockingbird", "Diary of a Wimpy Kid"],
    'passings':[3, 7, 2]
}

myVar = pd.DataFrame(myData)

print(myVar)
```

## Series in Pandas

```python
## Series are similar to just arrays of a column of a table 

cars = ["BMW", "Dodge", "Jeep", "Masserati"]

carData = pd.Series(cars)

print(carData)
```

## Labels 

```python
## If nothing else is specified in the index of a 0 will start

##This will print the first value of the series (expected output = BMW)
print(carData[0])


```

```python
# Example of creating labels

a = [1, 2, 3]

labelList = pd.Series(a, index = ["x", "y", "z"])

print(labelList)

##Print the value of a certain label 
print(labelList["z"])
```
