# Sobriety Model

## 1.1: Setup Data Augmentation and Preprocessing

```python
# TensorFlow & Keras for building the model
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# For visualization and evaluation
import matplotlib.pyplot as plt
import numpy as np
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)
```

## 1.2: Load the Training Images

```python
train_data = train_datagen.flow_from_directory(
    'Dataset/train',                   # path to folder with drowsy/ and natural/
    target_size=(224, 224),     # resize images to 224x224
    batch_size=32,
    class_mode='binary',        # binary classification: drowsy or natural
    subset='training'
)
```

## 1.3: Load the Validation Images

```python
val_data = train_datagen.flow_from_directory(
    'Dataset/test', 
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='validation'
) 
```

## 2.1: Start building the CNN (Convolutional Neural Network) 

```python
#Building the Convolutional Neurarl Network 

model = Sequential() 

```

## 2.2: Compile the model 

```python
model.compile(
    optimizer = 'adam', 
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
    
)
```

## 2.3: Train the model 

```python
history = model.fit(
    train_data,
    validation_data = val_data,
    epochs = 5
)
```

## 2.4 Save the Model

```python
model.save("sobriety_cnn_model.keras")

# How to load the model 
# from tensorflow.keras.models import load_model
# model = load_model("sobriety_cnn_model.h5")
```

## 3.1: Evaluate the Model

```python
loss, accuracy = model.evaluate(val_data)
print(f"Validation Accuracy: {accuracy:.2f}")

## Data is overfitting loss is ~ 0.59 - 0.62 
```

## 3.2: Plotting the ROC Curve 

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

y_true = []
y_pred = []

for images, labels in val_data:
    preds = model.predict(images)
    y_true.extend(labels)  # <-- FIXED this line
    y_pred.extend(preds)

y_true = np.array(y_true)
y_pred = np.array(y_pred).flatten()

# If labels are one-hot encoded, convert to class indices
if y_true.ndim > 1:
    y_true = np.argmax(y_true, axis=1)

# Compute ROC curve and AUC
fpr, tpr, thresholds = roc_curve(y_true, y_pred)
roc_auc = auc(fpr, tpr)

# Plotting
plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f"ROC curve (AUC = {roc_auc:.2f})")
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Receiver Operating Characteristic")
plt.legend(loc="lower right")
plt.grid()
plt.show()
```
